```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
print("Length of student_info:", len(student_info))
```

```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
student_info.clear()
print("Student Info after clear:", student_info)
```

```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
print("Name:", student_info.get("name"))
print("GPA:", student_info.get("GPA", "Not Available"))
```

```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
age = student_info.pop("age")
print("Popped Age:", age)
print("Student Info after pop:", student_info)
```

```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
last_item = student_info.popitem()
print("Popped Item:", last_item)
print("Student Info after popitem:", student_info)
```

```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"} 
keys = student_info.keys()
print("Keys:", keys)
```

```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
values = student_info.values()
print("Values:", values)
```

```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
items = student_info.items()
print("Items:", items)
```

```python
student_info = {"name": "John Doe", "age": 20, "grade": "A"} 
student_copy = student_info.copy()
print("Original Dictionary:", student_info)
print("Copied Dictionary:", student_copy)
```

```python
student_info = {"name": "John Doe", "age": 20}
extra_info = {"grade": "A", "GPA": 3.9} 
student_info.update(extra_info)
print("Updated Student Info:", student_info)
```
